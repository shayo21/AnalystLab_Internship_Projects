### DATASET DESCRIPTION FOR ANALYSIS

1.**E-commerce Transaction Dataset (OnlineRetail.csv)**
This is a dataset containing  all the transactions occurring between December 1, 2010, and December 9, 2011, for a UK-based and registered non-store online retail company. The company mainly sells unique all-occasion gifts, and many of its customers are whoLesalers. It tracks every row that represent a single product purchased within an order. It also tracks the invoice number, the specific product (stock code and description), the quantity purchased, the date/time of the transaction, the unit price, the unique customer ID, and the customer's country.

2. **Netflix Movies & TV Shows Dataset (netflix_titles.csv)**

This dataset consists of a comprehensive catalog of all the movies and TV shows available on the Netflix platform (as of late 2021), which indicate row represented in a single piece of content (a movie or a TV series). It details the content's unique ID, type (Movie or TV Show), title, director, cast members, country of production, the exact date it was added to the platform, original release year, age rating (e.g., TV-MA, PG-13), duration (in minutes or seasons), genres/categories, and a short text synopsis.

# Identifying important features in the dataset

During the analysis, some features were identified from the Netflix and Online Retail dataset. Numerical column for Netflix is release_year, while that of Retail is Quantity and unit price. Categorical columns are Country, type, and rating for Netflix.

Primary keys are unique keys in a dataset that helps to identify every single record or row. The primary keys for this analysis are show_id for Netflix, Invoice No and Stock Code for Retail.

## TASK 3: Exploratory Data Analysis (EDA)

### EDA for E-commerce

**1. Summary Statistics:**

Description such as mean, median, min, max, and standard deviation will be carried out for numerical columns ('Quantity' and 'UnitPrice').

**2. Exploring Data Patterns:**

To analyze financial patterns, 'Revenue'('UnitPrice' * 'Quantity') column will be created to explore:

* Top selling products by quantity
* Highest revenue generating countries
* Monthly sales trends
* Top spending customers
* Most purchased products by revenue

```python
# ---1. Summary statistics ---
print("---E-commerce Summary Statistics ---")
display(cleaned_retail.describe())

# create a revenue column for generating important insights
cleaned_retail['Revenue'] = cleaned_retail['Quantity'] * cleaned_retail['UnitPrice']

# ---2. Data Patterns
print("/n1. Top 5 selling product(By Quantity):")
print(cleaned_retail.groupby('Description')['Quantity'].sum().nlargest(5))

print("/n2. Top 5 Revenue Generating Countries:")
print(cleaned_retail.groupby('Country')['Revenue'].sum().nlargest(5))

print("/n3. Top 5 Monthly Sales Trends:")

# Transforming column with coercion to be a Datetime object
cleaned_retail['InvoiceDate'] = pd.to_datetime(cleaned_retail['InvoiceDate'])

# Extract month and year for clarity
cleaned_retail['MonthYear'] = cleaned_retail['InvoiceDate'].dt.to_period('M')
print(cleaned_retail.groupby('MonthYear')['Revenue'].sum().nlargest(5))

print("/n4. Top 5 Customer by Consumer Behaviour:")
print(cleaned_retail.groupby('CustomerID')['Revenue'].sum().nlargest(5))

print("/n5. Top 5 Products by Revenue:")
print(cleaned_retail.groupby('Description')['Revenue'].sum().nlargest(5))


```

## TASK 4: DATA VISUALIZATION

Below are 5 visualizations for exploring purchasing patterns, revenue generation, and consumer behavior for within the E-commerce dataset.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visualization styles
sns.set_theme(style= "whitegrid")
plt.figure(figsize=(18, 24))

# 1. Top selling products by Quantity (Bar Chart)
plt.subplot(3, 2, 1)
top_products = cleaned_retail.groupby('Description')['Quantity'].sum().nlargest(5)
sns.barplot(x=top_products.values, y=top_products.index, hue=top_products.index, palette='viridis')
plt.title('1. Top 5 selling products by Quantity')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Product')

# 2. Monthly sales trends (Line Chart)
plt.subplot(3, 2, 2)
monthly_sales = cleaned_retail.groupby('MonthYear')['Revenue'].sum()
monthly_sales.astype(float).plot(kind= 'line', marker= 'o', color= 'b', linewidth=2)
plt.title('2. Monthly Sales Revenue Trend')
plt.xlabel('Month-Year')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)

# 3. Highest revenue-generating countries - Excluding UK (Bar Chart)
plt.subplot(3, 2, 3)
top_countries = cleaned_retail[cleaned_retail['Country'] != 'United Kingdom'].groupby('Country')['Revenue'].sum(
).nlargest(5)
sns.barplot(x=top_countries.values, y=top_countries.index, hue=top_countries.index, palette='magma')
plt.title('3. Top 5 Countries by Revenue')
plt.xlabel('Total Revenue (€)')
plt.ylabel('Country')

# 4. Top spenders/ Consumers Behaviors (Bar Chart)
plt.subplot(3, 2, 4)
top_customers = cleaned_retail.groupby('CustomerID')['Revenue'].sum().nlargest(5)
sns.barplot(x=top_customers.index.astype(str), y=top_customers.values, hue= top_customers.index, palette='coolwarm')
plt.title('4. Total 5 Customers by Spending')
plt.xlabel('Customer ID')
plt.ylabel('Total Revenue (€)')

# 5. Distribution pf Unit Prices under 20 (Histogram)
plt.subplot(3, 2, 5)
sns.histplot(cleaned_retail[cleaned_retail['UnitPrice'] < 20]['UnitPrice'], bins=20, color='seagreen', kde=True)
plt.title('5. Distribution of Unit Prices (Items < €20)')
plt.xlabel('Unit Price(€)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()


```

**E-commerce Visualization Insights:**
1. **Top Products:** The "WORLD WAR 2 GLIDERS" and "JUMBO BAG RED RETROSPOT" are the highest volume items sold.
2. **Sales Trend:** Revenue steadily increases throughout the year, peaking massively in November due to holiday shopping.
3. **Top Countries:** While the UK is the main market, the Netherlands and EIRE (Ireland) are the highest revenue-generating international markets.
4. **Customer Behavior:** A few specific VIP customers (like ID 14646) spend vastly more than average, indicating strong B2B/wholesale relationships.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.gridspec as gridspec

# ==========================================
# 1. GLOBAL STYLING (The "Card" Look)
# ==========================================
plt.rcParams['figure.facecolor'] = '#E8EBEF' # Soft grey background for the whole dashboard
plt.rcParams['axes.facecolor'] = '#FFFFFF'   # White background for individual charts
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['axes.labelsize'] = 11

# Step 1: Ensure date columns and calculations are finalized
cleaned_retail['InvoiceDate'] = pd.to_datetime(cleaned_retail['InvoiceDate'])
cleaned_retail['MonthYear'] = cleaned_retail['InvoiceDate'].dt.to_period('M')
cleaned_retail['Revenue'] = cleaned_retail['Quantity'] * cleaned_retail['UnitPrice']

# Step 2: Calculate high-level summary metrics for KPI cards
total_revenue = cleaned_retail['Revenue'].sum()
total_quantities = cleaned_retail['Quantity'].sum()
unique_customers = cleaned_retail['CustomerID'].nunique()

# Step 3: Initialize a multi-row, multi-column dashboard grid
fig = plt.figure(figsize=(22, 16))
gs = gridspec.GridSpec(3, 3, figure=fig, height_ratios=[0.12, 1, 1])

# NEW: Unified Header Banner (Dark Blue)
fig.suptitle('E-COMMERCE EXECUTIVE PERFORMANCE DASHBOARD',
             fontsize=24, fontweight='heavy', color='white',
             bbox=dict(facecolor='#1A237E', edgecolor='none', boxstyle='square,pad=0.5'),
             y=0.96)

axes_list = [] # List to hold charts for final formatting

# --- SECTION 1: TOP ROW KPI CARDS ---
kpi_metrics = [
    (f"£{total_revenue:,.2f}", "Total Revenue Generated"),
    (f"{total_quantities:,}", "Total Items Sold"),
    (f"{unique_customers:,}", "Active Unique Customers")
]

for index, (value, metric_label) in enumerate(kpi_metrics):
    ax_kpi = fig.add_subplot(gs[0, index])
    ax_kpi.axis('off')

    # NEW: Draw a clean, white card with a subtle border for each metric
    card_border = plt.Rectangle((0.02, 0.05), 0.96, 0.90, fill=True, facecolor='white',
                                edgecolor='#CCCCCC', lw=1.5, transform=ax_kpi.transAxes)
    ax_kpi.add_patch(card_border)

    # Display the big numerical metric (Updated color)
    ax_kpi.text(0.5, 0.65, value, fontsize=32, fontweight='bold', ha='center', va='center', color='#1A237E')
    # Display the supporting context label below it
    ax_kpi.text(0.5, 0.3, metric_label, fontsize=14, fontweight='bold', ha='center', va='center', color='#666666')

# --- SECTION 2: CORE MACRO TRENDS ---
# Chart A: Monthly Sales Revenue Trend
ax_trend = fig.add_subplot(gs[1, 0:2])
monthly_sales = cleaned_retail.groupby('MonthYear')['Revenue'].sum()
monthly_sales.astype(float).plot(kind='line', marker='o', color='#2563eb', linewidth=3, ax=ax_trend)
ax_trend.set_title('Monthly Sales Revenue Growth Profile', pad=12)
ax_trend.set_ylabel('Revenue (£)')
ax_trend.set_xlabel('Timeline (Month-Year)')
ax_trend.tick_params(axis='x', rotation=45)
axes_list.append(ax_trend)

# Chart B: Top Countries by Revenue Excluding UK
ax_geo = fig.add_subplot(gs[1, 2])
top_countries = cleaned_retail[cleaned_retail['Country'] != 'United Kingdom'].groupby('Country')['Revenue'].sum().nlargest(5)
sns.barplot(x=top_countries.values, y=top_countries.index, hue=top_countries.index, palette='Blues_r', ax=ax_geo, legend=False)
ax_geo.set_title('Top 5 International Markets (Excl. UK)', pad=12)
ax_geo.set_xlabel('Total Revenue (£)')
ax_geo.set_ylabel('')
axes_list.append(ax_geo)

# --- SECTION 3: MICRO PERFORMANCE & INVENTORY ---
# Chart C: Top Selling Products by Volume
ax_prod = fig.add_subplot(gs[2, 0])
top_products = cleaned_retail.groupby('Description')['Quantity'].sum().nlargest(5)
sns.barplot(x=top_products.values, y=top_products.index, hue=top_products.index, palette='GnBu_r', ax=ax_prod, legend=False)
ax_prod.set_title('Top 5 Products by Volume Sold', pad=12)
ax_prod.set_xlabel('Total Units Sold')
ax_prod.set_ylabel('')
axes_list.append(ax_prod)

# Chart D: Top Spenders (Customer Concentration)
ax_cust = fig.add_subplot(gs[2, 1])
top_customers = cleaned_retail.groupby('CustomerID')['Revenue'].sum().nlargest(5)
sns.barplot(x=top_customers.index.astype(str), y=top_customers.values, hue=top_customers.index.astype(str), palette='YlGnBu_r', ax=ax_cust, legend=False)
ax_cust.set_title('Top 5 VIP High-Volume Customers', pad=12)
ax_cust.set_xlabel('Customer ID')
ax_cust.set_ylabel('Total Spend (£)')
axes_list.append(ax_cust)

# Chart E: Unit Price Distribution Profile
ax_price = fig.add_subplot(gs[2, 2])
sns.histplot(cleaned_retail[cleaned_retail['UnitPrice'] < 15]['UnitPrice'], bins=20, color='#0d9488', kde=True, ax=ax_price)
ax_price.set_title('Inventory Price Distribution Profile (< £15)', pad=12)
ax_price.set_xlabel('Unit Price (£)')
ax_price.set_ylabel('Inventory Count')
axes_list.append(ax_price)

# ==========================================
# 4. FINAL CLEANUP & FORMATTING LOOP
# ==========================================
for ax in axes_list:
    # Add subtle horizontal gridlines
    ax.yaxis.grid(True, color='#E0E0E0', linestyle='-', linewidth=1)
    ax.set_axisbelow(True) # Put grid behind bars/lines

    # Clean up the outer borders
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_color('#CCCCCC')
    ax.spines['bottom'].set_color('#CCCCCC')

# Step 4: Optimize spacing gaps across elements
plt.tight_layout()
plt.subplots_adjust(top=0.90, hspace=0.45, wspace=0.28) # Adjusted top and hspace slightly for better breathing room
plt.show()
```
